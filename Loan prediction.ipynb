Model Building

```python
import pandas as pd

# 1. Loan_ID hatao Drop column safely
df = df.drop(columns=['Loan_ID'], errors='ignore',axis=1)

# 2. Categorical → Numeric
df = pd.get_dummies(df, drop_first=True)

# 3. Missing values handle
df = df.fillna(df.mean(numeric_only=True))

# 4. X aur y define karo
x = df.drop('Loan_Status_Y', axis=1)
y = df['Loan_Status_Y']

# 5. Train test split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# 6. Models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "DecisionTree": DecisionTreeClassifier(),
    "RandomForest": RandomForestClassifier()
}

# 7. Train + Predict
from sklearn.metrics import accuracy_score

for name, model in models.items():
    model.fit(x_train, y_train)
    preds = model.predict(x_test)
    print(name, "Accuracy:", accuracy_score(y_test, preds))
```

```python
y
```

```python
Why Random Forest usually wins here

Loan approval data typically has:
Non-linear relationships (e.g., CIBIL score interacts with income)
Feature interactions (income assets dependents together matter)
Some noise and outliers


Random Forest handles all of this very well because:
It captures non-linear patterns
It reduces overfitting compared to a single tree
It works well even without perfect feature scaling
It usually gives higher accuracy and better generalization
```

```python
# Step 1: Split Features and Target 

x = df.drop('Loan_Status_Y', axis=1)
y = df['Loan_Status_Y']
```

```python
# Step 2: Train-Test Split

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

```python
# Step 3: Build Random Forest Model

from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier ( n_estimators = 200, max_depth=10, random_state = 39)
```

```python
rf_model.fit(x_train, y_train)
```

```python
# Step 4: Make Predictions

y_pred= rf_model.predict(x_test)
```

```python
# Step 5: Evaluate the Model

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accurancy:",accuracy_score(y_test,y_pred)*100,"%")
print("Classification Report:",classification_report(y_test,y_pred))
print("Confusion Matrix:",confusion_matrix(y_test,y_pred))
```

```python
# step:6 Feature Importance 

feature_importance =pd.Series(rf_model.feature_importances_,index =x.columns).sort_values(ascending=False)

feature_importance
```

```python
import matplotlib.pyplot as plt

feature_importance.plot(kind='bar')
plt.title("Top 10 Importance Features -Random Forest")
plt.ylabel("Importance Score")
plt.show()
```
